```python
movies_raw = pd.read_csv("/content/tmdb_5000_movies.csv")
credits_raw = pd.read_csv("/content/tmdb_5000_credits.csv")

print("Original datasets loaded successfully!")
print("Movies:", movies_raw.shape)
print("Credits:", credits_raw.shape)

# Select required credits columns
credits_clean = credits_raw[["movie_id", "cast", "crew"]].copy()

# Merge datasets
merged_movies = movies_raw.merge(
    credits_clean,
    left_on="id",
    right_on="movie_id",
    how="inner",
    validate="one_to_one"
)

# Select required columns
movies = merged_movies[
    ["id", "title", "overview", "genres", "keywords", "cast", "crew"]
].copy()

# Rename ID
movies.rename(columns={"id": "movie_id"}, inplace=True)

print("Merge completed successfully!")
print("Dataset shape:", movies.shape)
print("Columns:", movies.columns.tolist())
```

```python

print("Missing values in Movies dataset:")
print(movies.isnull().sum())
```

```python
print("Missing values in Credits dataset:")
print(credits.isnull().sum())
```

```python

# Remove rows with missing essential information
movies.dropna(
    subset=["movie_id", "title", "overview"],
    inplace=True
)

# Fill missing metadata
for column in ["genres", "keywords", "cast", "crew"]:
    movies[column] = movies[column].fillna("[]")

movies.reset_index(drop=True, inplace=True)

print("Missing values handled successfully!")
print(movies.isnull().sum())
print("\nCleaned dataset shape:", movies.shape)
```

```python

print("Dataset shape:", movies.shape)
print("Column names:", movies.columns.tolist())

display(movies.head())

```

```python

import ast

def extract_names(text):
    try:
        data = ast.literal_eval(text) if isinstance(text, str) else []
        return [item["name"] for item in data if "name" in item]
    except (ValueError, SyntaxError, TypeError):
        return []

movies["genres"] = movies["genres"].apply(extract_names)

print(movies[["title", "genres"]].head())
```

```python

movies["keywords"] = movies["keywords"].apply(extract_names)

print(movies[["title", "keywords"]].head())
```

```python

def extract_cast(text):
    try:
        data = ast.literal_eval(text) if isinstance(text, str) else []
        return [item["name"] for item in data[:3] if "name" in item]
    except (ValueError, SyntaxError, TypeError):
        return []

movies["cast"] = movies["cast"].apply(extract_cast)

print("Top three actors extracted successfully!")
display(movies[["title", "cast"]].head())
```

```python

def extract_director(text):
    try:
        data = ast.literal_eval(text) if isinstance(text, str) else []
        for item in data:
            if item.get("job") == "Director":
                return item["name"]
        return ""
    except (ValueError, SyntaxError, TypeError):
        return ""

movies["director"] = movies["crew"].apply(extract_director)

print("Directors extracted successfully!")
display(movies[["title", "director"]].head())
```

```python

# Remove spaces from genres, keywords and cast
for column in ["genres", "keywords", "cast"]:
    movies[column] = movies[column].apply(
        lambda x: [str(i).replace(" ", "") for i in x]
    )

# Convert director names into lists
movies["director"] = movies["director"].apply(
    lambda x: [x.replace(" ", "")] if isinstance(x, str) and x else []
)

print("Spaces removed successfully!")
display(movies[["title", "genres", "keywords", "cast", "director"]].head())


```

```python

# Convert all features into lists of words safely

def make_list(value):
    if isinstance(value, list):
        return value

    if isinstance(value, str):
        return value.lower().split()

    return []

# Ensure every feature is a list
feature_columns = [
    "overview",
    "genres",
    "keywords",
    "cast",
    "director"
]

for column in feature_columns:
    movies[column] = movies[column].apply(make_list)

# Combine all features into tags
movies["tags"] = (
    movies["overview"]
    + movies["genres"]
    + movies["keywords"]
    + movies["cast"]
    + movies["director"]
)

print("Tags created successfully!")

display(movies[["title", "tags"]].head())
```

```python

# Keep the required columns
final_movies = movies[["movie_id", "title", "tags"]].copy()

# Convert tag lists into strings
final_movies["tags"] = final_movies["tags"].apply(
    lambda x: " ".join(x).lower()
)

print("Final dataset created successfully!")
print("Dataset shape:", final_movies.shape)

display(final_movies.head())
```

```python

from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(
    max_features=5000,
    stop_words="english"
)

print("CountVectorizer initialized successfully!")
```

```python

vectors = cv.fit_transform(final_movies["tags"]).toarray()

print("Text vectorization completed successfully!")
print("Vector matrix shape:", vectors.shape)
```

```python

from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)

print("Cosine similarity calculated successfully!")
print("Similarity matrix shape:", similarity.shape)
```

```python

def recommend(movie_name):
    movie_name = movie_name.strip().lower()

    matches = final_movies[
        final_movies["title"].str.lower() == movie_name
    ]

    if matches.empty:
        print(f"Movie '{movie_name}' was not found.")
        print("Try a movie title from final_movies['title'].")
        return

    movie_index = matches.index[0]

    distances = list(enumerate(similarity[movie_index]))
    distances = sorted(
        distances,
        key=lambda item: item[1],
        reverse=True
    )

    print(f"Movies similar to {final_movies.iloc[movie_index]['title']}:\n")

    count = 0
    for index, score in distances[1:]:
        print(f"{final_movies.iloc[index]['title']} (similarity: {score:.3f})")
        count += 1
        if count == 5:
            break
```

```python
from collections import Counter

genre_counts = Counter(
    genre
    for genres in movies["genres"]
    for genre in genres
)

print("Available movie genres:\n")

for genre, count in genre_counts.most_common():
    print(f"{genre}: {count} movies")
```

```python
def recommend_by_genre(genre, top_n=5):

    genre = genre.strip().lower()

    # Find movies belonging to the selected genre
    filtered_movies = movies[
        movies["genres"].apply(
            lambda genres: genre in [g.lower() for g in genres]
        )
    ]

    if filtered_movies.empty:
        print("No movies found for this genre.")
        return

    print(f"\nTop {top_n} {genre.title()} Movies:\n")

    display(
        filtered_movies[
            ["movie_id", "title", "genres"]
        ].head(top_n)
    )
```

```python
recommend_by_genre("Action")
```
